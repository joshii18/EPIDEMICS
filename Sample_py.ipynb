```python

# ============================================================
#     B.TECH MATHEMATICS MINI PROJECT
#     EPIDEMIC SPREAD OVER N WEEKS (C1)
# ============================================================
#
# Mathematical model:
#       X(n+1) = A X(n)
#       X(n)   = A^n X(0)
#
# We use Cayley-Hamilton theorem to calculate A^n efficiently.
#
# ASSUMPTION:
# This is a LINEAR transition-matrix model.
# Columns of A represent the current state and rows represent
# the next state.
#
# State order:
# S = Susceptible
# I = Infected
# R = Recovered
#
# ============================================================

# ---------------------------
# 1. IMPORT LIBRARIES
# ---------------------------

import numpy as np
import pandas as pd
import sympy as sp
import matplotlib.pyplot as plt

from IPython.display import display, Markdown

plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["axes.grid"] = True


# ============================================================
# 2. USER CONFIGURATION
# ============================================================

N_WEEKS = 52

# Initial population:
# [Susceptible, Infected, Recovered]

INITIAL_POPULATION = np.array([
    990,
    10,
    0
], dtype=float)

# Weekly transition matrix
#
#          Current:    S       I       R
#
# Next S              0.94    0       0
# Next I              0.06    0.90    0
# Next R              0       0.10    1.00
#
# Each column adds to 1.

A = np.array([
    [0.94, 0.00, 0.00],
    [0.06, 0.90, 0.00],
    [0.00, 0.10, 1.00]
], dtype=float)

STATE_NAMES = [
    "Susceptible",
    "Infected",
    "Recovered"
]


# ============================================================
# 3. DISPLAY INPUTS
# ============================================================

print("=" * 65)
print("          EPIDEMIC SPREAD OVER N WEEKS")
print("=" * 65)

print("\nPROJECT INPUTS")
print("-" * 65)

print(f"Number of weeks: {N_WEEKS}")

print("\nInitial Population:")
for name, value in zip(STATE_NAMES, INITIAL_POPULATION):
    print(f"{name:15s}: {value:,.0f}")

print("\nWeekly Transition Matrix A:")
print(A)


# ============================================================
# 4. VALIDATE MODEL
# ============================================================

if A.shape != (3, 3):
    raise ValueError("Transition matrix A must be 3 × 3.")

if INITIAL_POPULATION.shape != (3,):
    raise ValueError("Initial population must contain 3 values.")

if np.any(A < 0):
    raise ValueError("Transition probabilities cannot be negative.")

if np.any(INITIAL_POPULATION < 0):
    raise ValueError("Population cannot be negative.")

if not np.allclose(A.sum(axis=0), 1):
    raise ValueError(
        "Each column of A must sum to 1."
    )

print("\n✓ Model validation successful.")


# ============================================================
# 5. CREATE EXACT SYMPY MATRIX
# ============================================================

A_sym = sp.Matrix([
    [sp.Rational(str(x)) for x in row]
    for row in A
])

X0_sym = sp.Matrix([
    sp.Rational(str(x))
    for x in INITIAL_POPULATION
])

I_matrix = sp.eye(3)

print("\nExact symbolic matrix A:")
display(A_sym)


# ============================================================
# 6. CHARACTERISTIC EQUATION
# ============================================================

lam = sp.symbols("lambda")

char_poly = A_sym.charpoly(lam)

characteristic_equation = sp.expand(
    char_poly.as_expr()
)

print("\n" + "=" * 65)
print("CHARACTERISTIC EQUATION")
print("=" * 65)

print("\np(λ) =")
display(characteristic_equation)

print("\nFactored form:")
display(sp.factor(characteristic_equation))

coefficients = char_poly.all_coeffs()

print("\nCoefficients:")
print(coefficients)


# ============================================================
# 7. CAYLEY-HAMILTON THEOREM
# ============================================================

print("\n" + "=" * 65)
print("CAYLEY-HAMILTON THEOREM")
print("=" * 65)

print("""
Cayley-Hamilton theorem states:

Every square matrix satisfies its own
characteristic equation.
""")

a3, a2, a1, a0 = coefficients

# Characteristic equation:
#
# A^3 + a2 A^2 + a1 A + a0 I = 0

print("Therefore:")

display(
    sp.Eq(
        A_sym**3
        + a2 * A_sym**2
        + a1 * A_sym
        + a0 * I_matrix,
        sp.zeros(3)
    )
)

print("\nHence:")

display(
    sp.Eq(
        A_sym**3,
        -a2 * A_sym**2
        -a1 * A_sym
        -a0 * I_matrix
    )
)


# ============================================================
# 8. VERIFY CAYLEY-HAMILTON
# ============================================================

CH_result = sp.simplify(
    A_sym**3
    + a2 * A_sym**2
    + a1 * A_sym
    + a0 * I_matrix
)

print("\nCayley-Hamilton verification:")

if CH_result == sp.zeros(3):
    print("✓ VERIFIED: A satisfies its characteristic equation.")
else:
    print("✗ Verification failed.")


# ============================================================
# 9. CAYLEY-HAMILTON POWER FUNCTION
# ============================================================

def matrix_power_cayley_hamilton(A_sym, n):
    """
    Calculate A^n using Cayley-Hamilton theorem.

    x^n is reduced modulo the characteristic polynomial.

    Therefore:

        A^n = c2 A^2 + c1 A + c0 I

    for a 3 × 3 matrix.
    """

    if n < 0:
        raise ValueError("Power must be non-negative.")

    if n == 0:
        return sp.eye(A_sym.rows), sp.Integer(1)

    x = sp.symbols("x")

    characteristic = A_sym.charpoly(x).as_expr()

    # Reduce x^n modulo characteristic polynomial
    remainder = sp.rem(
        x**n,
        characteristic,
        domain=sp.QQ
    )

    remainder_poly = sp.Poly(remainder, x)

    result = sp.zeros(A_sym.rows)

    for (power,), coefficient in remainder_poly.terms():

        result += coefficient * (
            A_sym ** power
        )

    return sp.simplify(result), remainder


# ============================================================
# 10. CALCULATE A^52
# ============================================================

A52_sym, remainder_52 = (
    matrix_power_cayley_hamilton(
        A_sym,
        N_WEEKS
    )
)

print("\n" + "=" * 65)
print(f"CAYLEY-HAMILTON CALCULATION OF A^{N_WEEKS}")
print("=" * 65)

print(f"\nx^{N_WEEKS} reduced modulo the characteristic polynomial:")

display(remainder_52)

print(f"\nTherefore A^{N_WEEKS} =")

display(A52_sym)


# ============================================================
# 11. VERIFY AGAINST NUMPY
# ============================================================

A52_CH = np.array(
    A52_sym,
    dtype=float
)

A52_numpy = np.linalg.matrix_power(
    A,
    N_WEEKS
)

difference = np.max(
    np.abs(A52_CH - A52_numpy)
)

print("\nVerification using NumPy:")
print(
    f"Maximum difference = {difference:.3e}"
)

if np.allclose(
    A52_CH,
    A52_numpy,
    atol=1e-10
):
    print(
        "✓ Cayley-Hamilton result matches NumPy."
    )
else:
    print(
        "✗ Results do not match."
    )


# ============================================================
# 12. POPULATION AFTER 52 WEEKS
# ============================================================

X0 = INITIAL_POPULATION

X52 = A52_CH @ X0

print("\n" + "=" * 65)
print(f"POPULATION AFTER {N_WEEKS} WEEKS")
print("=" * 65)

for name, value in zip(
    STATE_NAMES,
    X52
):
    print(
        f"{name:15s}: {value:,.2f}"
    )

print(
    f"{'Total':15s}: {X52.sum():,.2f}"
)


# ============================================================
# 13. CALCULATE EPIDEMIC EVOLUTION
# ============================================================

weekly_results = []

for week in range(N_WEEKS + 1):

    A_n_sym, _ = (
        matrix_power_cayley_hamilton(
            A_sym,
            week
        )
    )

    A_n = np.array(
        A_n_sym,
        dtype=float
    )

    X_n = A_n @ X0

    weekly_results.append([
        week,
        X_n[0],
        X_n[1],
        X_n[2],
        X_n.sum()
    ])


results_df = pd.DataFrame(
    weekly_results,
    columns=[
        "Week",
        "Susceptible",
        "Infected",
        "Recovered",
        "Total"
    ]
)


# ============================================================
# 14. DISPLAY WEEKLY TABLE
# ============================================================

print("\n" + "=" * 65)
print("EPIDEMIC EVOLUTION")
print("=" * 65)

display(
    results_df.round(2)
)


# ============================================================
# 15. SELECT IMPORTANT WEEKS
# ============================================================

selected_weeks = [
    0,
    1,
    2,
    4,
    8,
    12,
    20,
    30,
    40,
    52
]

selected_df = results_df[
    results_df["Week"].isin(
        selected_weeks
    )
].copy()

print("\nSelected weeks:")
display(
    selected_df.round(2)
)


# ============================================================
# 16. FIND PEAK INFECTION
# ============================================================

peak_index = (
    results_df["Infected"].idxmax()
)

peak_week = int(
    results_df.loc[
        peak_index,
        "Week"
    ]
)

peak_infected = (
    results_df.loc[
        peak_index,
        "Infected"
    ]
)

print("\n" + "=" * 65)
print("EPIDEMIC PEAK")
print("=" * 65)

print(
    f"Peak infected population : "
    f"{peak_infected:.2f}"
)

print(
    f"Peak occurs at week      : "
    f"{peak_week}"
)


# ============================================================
# 17. STEADY STATE
# ============================================================

print("\n" + "=" * 65)
print("STEADY STATE")
print("=" * 65)

print("""
At steady state:

X* = A X*

Therefore:

(A - I)X* = 0
""")

steady_vectors = (
    A_sym - I_matrix
).nullspace()

if len(steady_vectors) == 0:

    print(
        "No steady-state vector with "
        "eigenvalue 1 was found."
    )

else:

    steady_vector = (
        steady_vectors[0]
    )

    # Normalize total population to 1
    steady_vector = (
        steady_vector /
        sum(steady_vector)
    )

    total_population = (
        INITIAL_POPULATION.sum()
    )

    steady_population = np.array([
        float(value * total_population)
        for value in steady_vector
    ])

    print("Steady-state proportions:\n")

    for name, value in zip(
        STATE_NAMES,
        steady_vector
    ):

        print(
            f"{name:15s}: "
            f"{float(value):.6f}"
        )

    print("\nSteady-state population:\n")

    for name, value in zip(
        STATE_NAMES,
        steady_population
    ):

        print(
            f"{name:15s}: "
            f"{value:.2f}"
        )


# ============================================================
# 18. DISTANCE FROM STEADY STATE
# ============================================================

distance = np.linalg.norm(
    X52 - steady_population
)

print(
    f"\nDistance of week {N_WEEKS} "
    f"from steady state: "
    f"{distance:.4f}"
)


# ============================================================
# 19. CHART 1 — SIR EVOLUTION
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    results_df["Week"],
    results_df["Susceptible"],
    linewidth=3,
    label="Susceptible (S)"
)

plt.plot(
    results_df["Week"],
    results_df["Infected"],
    linewidth=3,
    label="Infected (I)"
)

plt.plot(
    results_df["Week"],
    results_df["Recovered"],
    linewidth=3,
    label="Recovered (R)"
)

plt.title(
    "Epidemic Spread Over N Weeks",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Week",
    fontsize=13
)

plt.ylabel(
    "Population",
    fontsize=13
)

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 20. CHART 2 — INFECTED POPULATION
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    results_df["Week"],
    results_df["Infected"],
    marker="o",
    markersize=3,
    linewidth=2
)

plt.axvline(
    peak_week,
    linestyle="--",
    linewidth=2,
    label=f"Peak: Week {peak_week}"
)

plt.scatter(
    peak_week,
    peak_infected,
    s=100,
    zorder=5
)

plt.annotate(
    f"Peak ≈ {peak_infected:.1f}",
    xy=(
        peak_week,
        peak_infected
    ),
    xytext=(
        peak_week + 3,
        peak_infected
    ),
    arrowprops=dict(
        arrowstyle="->"
    )
)

plt.title(
    "Infected Population Over Time",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel("Week")
plt.ylabel("Number of Infected People")

plt.legend()

plt.tight_layout()

plt.show()


# ============================================================
# 21. CHART 3 — POPULATION DISTRIBUTION
# ============================================================

plt.figure(figsize=(12, 6))

plt.stackplot(
    results_df["Week"],
    results_df["Susceptible"],
    results_df["Infected"],
    results_df["Recovered"],
    labels=[
        "Susceptible",
        "Infected",
        "Recovered"
    ],
    alpha=0.8
)

plt.title(
    "Population Distribution Over 52 Weeks",
    fontsize=17,
    fontweight="bold"
)

plt.xlabel("Week")
plt.ylabel("Population")

plt.legend(
    loc="upper right"
)

plt.tight_layout()

plt.show()


# ============================================================
# 22. FINAL PROJECT SUMMARY
# ============================================================

print("\n")
print("=" * 65)
print("                 PROJECT SUMMARY")
print("=" * 65)

print("""
Topic:
Epidemic Spread Over N Weeks

Mathematical Model:
    X(n+1) = A X(n)

Therefore:
    X(n) = A^n X(0)

Main Mathematical Concept:
    Cayley-Hamilton Theorem

Characteristic Equation:
    det(lambda I - A) = 0

Using Cayley-Hamilton:
    A^n can be reduced to a combination
    of I, A and A^2.

For one year:
    n = 52

Therefore:
    X(52) = A^52 X(0)

Steady State:
    X* = A X*

or:
    (A - I)X* = 0
""")

print("\nFinal 52-week population:")

for name, value in zip(
    STATE_NAMES,
    X52
):

    print(
        f"{name:15s}: "
        f"{value:,.2f}"
    )

print(
    f"\nPeak infection: "
    f"{peak_infected:.2f} "
    f"people in week {peak_week}"
)

print(
    "\n✓ COMPLETE PROJECT EXECUTED SUCCESSFULLY"
)

print("=" * 65)
```
